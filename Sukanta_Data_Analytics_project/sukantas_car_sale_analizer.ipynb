```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
df=pd.read_csv("sale.csv")
df.head()
df.shape
df.info()
df.describe()
df.isnull().sum()
# step 10 analze the data
df.groupby("Fuel_Type")["sale"].sum()
df.groupby("Car_Name")["Year"].sum().sort_values(ascending=False)
print(f"Total sale:{df["sale"].sum()} \n Avrage sale:{df['sale'].mean()}\n Maximun sale:{df['sale'].max()}\n minimun sale:{df['sale'].min()}")

#visualize
#Analyze fule type
fuel_sales = df.groupby("Fuel_Type")["sale"].sum()

print(fuel_sales)

fuel_sales.plot(kind="bar")

plt.title("Total Sales by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Total Sale")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


#Analyze transmission type
transmission_sales = df.groupby("Transmission")["sale"].sum()

print(transmission_sales)

transmission_sales.plot(kind="bar")

plt.title("Total Sales by Transmission")
plt.xlabel("Transmission")
plt.ylabel("Total Sale")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


#sale by company....
company_sales = df.groupby("Company")["sale"].sum().sort_values(ascending=False)

print(company_sales)

company_sales.plot(kind="bar")

plt.title("Total Sales by Company")
plt.xlabel("Company")
plt.ylabel("Total Sale")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


#analyze the car
plt.scatter(df["Year"], df["sale"])

plt.title("Car Year vs Sale")
plt.xlabel("Year")
plt.ylabel("Sale")

plt.tight_layout()
plt.show()


#engine size vs sale
plt.scatter(df["Engine_CC"], df["sale"])

plt.title("Engine Size vs Sale")
plt.xlabel("Engine CC")
plt.ylabel("Sale")

plt.tight_layout()
plt.show()


#milage
plt.scatter(df["Mileage"], df["sale"])

plt.title("Mileage vs Sale")
plt.xlabel("Mileage")
plt.ylabel("Sale")

plt.tight_layout()
plt.show()


correlation = df[["Year", "sale", "Mileage", "Engine_CC"]].corr()

print(correlation)
sns.heatmap(correlation, annot=True)

plt.title("Correlation Between Numerical Variables")
plt.tight_layout()
plt.show()
### show all graphs
```
